```python
#series- A pandas series is 1-D array like object that can hold any data type
import pandas as pd
data=[1,2,3,4,5,9]
series=pd.Series(data)
print(series)
print(type(series))
```

```python
#create series from dictionary elements
#here key of dictionary become indexes
data={'a':1, 'b':2, 'c': 3, 'd':4}
series_dict=pd.Series(data)
print(series_dict)
```

```python
data =[3,5,7,9]
indexes=['b','h','a','o']
pd.Series(data, index=indexes)
```

```python
#Dataframe
#create a dataframe from dict of list
data={
    'Name': ['hiba', 'naba', 'alham'],
    'age': [21,20,16],
    'city': ['bgp', 'lmp', 'lko']
}
df=pd.DataFrame(data)
print(df)
print(type(df))
```

```python
import numpy as np
np.array(data)
```
