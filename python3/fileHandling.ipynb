```python
# Read a whole file
with open('ex.txt', 'r') as file:
    content = file.read()
    print(content) 
```

```python
#read a file line by line
with open('ex.txt', 'r') as file:
    for line in file:
        #print(line)
        print(line.strip())  # strip() removes the newline character at the end of each line
```

```python
#writing a file(overwriting)
with open('ex.txt', 'w') as file:
    file.write("This is a new line.\n")
    file.write("This will overwrite the existing content.\n")
```

```python
## Write to a file without overwriting (appending)
with open('ex.txt', 'a') as file:
    file.write("This line will be appended to the file.\n")
    file.write("This is another appended line.\n")
```

```python
# writing list of lines to a file
lines = ["First line\n", "Second line\n", "Third line\n"]
with open('ex.txt', 'a') as file:
    file.writelines(lines)
```

```python
##binary files
data = b'\x00\x01\x02\x03\x04\x05'
with open('ex.bin', 'wb') as file:
    file.write(data)
```

```python
#read the comtent from source text file and write it to destination text file 
with open('ex.txt', 'r') as source_file:
    content = source_file.read()

with open('dest_copy.txt', 'w') as destination_file:
    destination_file.write(content)    
```

```python
#writing and then reading a file
with open('example.txt', 'w+') as file:
    file.write("This is a new line.\n")
    file.write("This will overwrite the existing content.\n")

    #move the file pointer to the beginning of the file before reading
    file.seek(0)
    content = file.read()
    print(content)
```

```python
import os
path='ex.txt'

if os.path.isfile(path):
    print(f"The path '{path}' is a file.")
elif os.path.isdir(path):
    print(f"The path '{path}' is a directory.")
else:
    print(f"The path '{path}' does not exist or is neither a file nor a directory.")    
```
