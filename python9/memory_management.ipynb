```python
#best practices of memory management
#1- Use local variables
#2- avoid circular references(a==b and b==a) and use generators
#3- explicitly delete objects- use delete statement to del objs.
#4- use mem profiling tools like tracemalloc and memory_profiler to identify memory leaks and optimize mem usage.

```

```python
import gc
class MyObject:
    def __init__(self,name):
        self.name=name
        print(f"Object {self.name} created!")

    def __del__(self):
        print(f"Onject {self.name} deleted!")  

#creating circular reference

obj1=MyObject("obj1")
obj2=MyObject("obj2")
obj1.ref=obj2
obj2.ref=obj1

del obj1
del obj2


#due to circular ref it won't get deleted


          
```

```python


import gc
class MyObject:
    def __init__(self,name):
        self.name=name
        print(f"Object {self.name} created!")

    def __del__(self):
        print(f"Onject {self.name} deleted!")  

#handling and creating circular reference

obj1 = MyObject("obj1")
obj2 = MyObject("obj2")
obj1.ref = obj2
obj2.ref = obj1

del obj1
del obj2

#manually trigger the garbage collection

print(gc.collect())
```

```python
#printing all the collected garbage
print(f"Garbage collected objects: {gc.garbage}")

```

```python
#generators for memory efficiency: it allows to produce one item at a time using mem efficiently
def generate_num(n):
    for i in range(n):
        yield i
#using the generator

for num in generate_num(1000000):
    print(num)
    if num>50:
        break
        
```

```python
##profiling memory usage with tracemalloc
import tracemalloc

def create_list():
    return [i for i in range(1000)]

def main():
    tracemalloc.start()

    create_list()

    snapshot = tracemalloc.take_snapshot()
    top_stats = snapshot.statistics('lineno')

    print("[Top 10 ]")
    for stat in top_stats[::]:
        print(stat)
```

```python
main()

#this shows entire memory profiling
```
