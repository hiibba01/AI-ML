```python
## decorators are functions which take another function as an argument and extends the behaviour of the latter function without explicitly modifying it.

def main_welcome(func):
    
    def sub_welcome():
        print("WELCOME BACKKK")
        print("welcome everyone!")
        func()
    return sub_welcome()    
```

```python
def course_intro():
    print("This is a python course")
course_intro()    
```

```python
main_welcome(course_intro)
```

```python
@main_welcome
def course_intro():
    print("This is a python course")
```

```python
##another example
def my_decorator(func):
    def wrapper():
        print("Something is happening before the function is called.")
        func()
        print("Something is happening after the function is called.")
    return wrapper
```

```python
@my_decorator
def say_hello():
    print("Hello!")
say_hello()    
```

```python
# decorators with arguments
def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator    
```

```python
@repeat(3)
def say_hello():
    print("Hello!")
```

```python
say_hello()
```
