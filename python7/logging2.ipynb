```python
import logging
## create a logger for module1
logger1=logging.getLogger("module1")
logger1.setLevel(logging.DEBUG)

#create a logger for module2
logger2=logging.getLogger("module2")
logger2.setLevel(logging.WARNING)


#configure logging setting
logging.basicConfig(
    filename='app.log',  #app.log file will be created and content will be written in it
    filemode='w',
    level=logging.DEBUG,
    format='%(asctime)s-%(name)s-%(levelname)s-%(message)s',
    datefmt='%Y-%m-%d %H:%M:%S',
    force=True

)

```

```python
#log message with different loggers
logger1.debug("this is debug msg for module1")
logger2.warning("this is warning msg for module2")
logger2.error({"this is an error msg"})
```
