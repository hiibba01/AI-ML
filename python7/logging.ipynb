```python
import logging
```

```python
#configure the basic logging settings
logging.basicConfig(level=logging.DEBUG)

#LOGGING MESSAGES with different severity levels
logging.debug("This is a debug message")
logging.info("This is an info message")
logging.warning("A warning msg")
logging.error("An error msg")
logging.critical("A critical msg")
```

```python
#configuring logging
import logging

logging.basicConfig(
    level=logging.DEBUG,
    format='%(asctime)s-%(name)s-%(levelname)s-%(message)s',
    datefmt='%Y-%m-%d %H:%M:%S',
    force=True

)

#LOGGING MESSAGES with different severity levels
logging.debug("This is a debug message")
logging.info("This is an info message")
logging.warning("A warning msg")
logging.error("An error msg")
logging.critical("A critical msg")
```

```python
#configuring logging
import logging

logging.basicConfig(
    filename='app.log',  #app.log file will be created and content will be written in it
    filemode='w',
    level=logging.DEBUG,
    format='%(asctime)s-%(name)s-%(levelname)s-%(message)s',
    datefmt='%Y-%m-%d %H:%M:%S',
    force=True

)

#LOGGING MESSAGES with different severity levels
logging.debug("This is a debug message")
logging.info("This is an info message")
logging.warning("A warning msg")
logging.error("An error msg")
logging.critical("A critical msg")
```

```python
logging.debug("This is a debug message 2")
logging.info("This is an info message 2")
logging.warning("A warning msg 2")
logging.error("An error msg 2")
logging.critical("A critical msg 2") 
#now all these msgs will be executed in app.log file
```
