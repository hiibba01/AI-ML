```python
import sqlite3
```

```python
#connect to sqlite database
connection=sqlite3.connect('example.db')
connection
cursor=connection.cursor()
```

```python
#create a table
cursor.execute('''
Create Table If Not Exists employees(
    id Integer Primary Key,
    name Text Not Null,
    age Integer,
    department text
    )
''')

#commit the change
connection.commit()

```

```python
cursor.execute('''
Select * from employees
''')

#this displays all records
### we can also write-
#SELECT * FROM employees WHERE age > 80;
#This means top select all the employees whose age is more than 80
```

```python
#insert the data in sqlite table
cursor.execute('''
    Insert Into employees(name,age,department)
        values('Hiba',20,'Data Scientist')
''')

cursor.execute('''
    Insert Into employees(name,age,department)
        values('Naba',19,'AI & ML')
''')

cursor.execute('''
    Insert Into employees(name,age,department)
        values('Alham',16,'Finance')
''')
#commiting the changes
connection.commit()
```

```python
#Query the data from table
cursor.execute('Select * from employees')
rows=cursor.fetchall()

#print the queried data

for row in rows:
    print(row)
```

```python
#update the table in the table
cursor.execute('''
Update employees
Set age=30
where name='Hiba'
''')

connection.commit()


#Query the data from table
cursor.execute('Select * from employees')
rows=cursor.fetchall()

#print the queried data

for row in rows:
    print(row)
```

```python
#deleting data fom table
cursor.execute('''
Delete from employees where name='Naba'
''')
```

```python
#working with sales data

connection=sqlite3.connect('sales_data.db')
cursor=connection.cursor()

#create a table for sales data
cursor.execute('''
CREATE TABLE IF NOT EXISTS sales(
    id INT PRIMARY KEY AUTOINCREMENT,
    date TEXT NOT NULL,
    product TEXT NOT NULL,
    sales INTEGER,
    region TEXT
)
''')

sales_data = [
    ('2026-01-01', 'Product1', 100, 'North'),
    ('2026-02-08', 'Product2',200, 'South'),
    ('2026-03-04', 'Product3', 400, 'East'),
    ('2026-04-06', 'Product4', 700, 'West')

]


cursor.executemany('''
INSERT INTO sales(date,product,sales,region)
    values(?,?,?,?)
''',sales_data)

connection.commit()

print("Data inserted successfully!")


```
