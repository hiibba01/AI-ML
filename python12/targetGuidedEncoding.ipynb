```python
#Target Guided Ordinal Encoding
```

```python
import pandas as pd
df = pd.DataFrame({
    'city': ['New York', 'Los Angeles', 'Chicago', 'New York', 'Paris', 'Los Angeles'],
    'price': [100, 200, 150, 300, 250, 400]
})
```

```python
df
```

```python
#consider price as the target variable and city as the categorical variable. We will perform target guided ordinal encoding on the city variable based on the price variable.
```

```python
mean_price=df.groupby('city')['price'].mean().to_dict()
```

```python
mean_price
```

```python
df['city_encoded'] = df['city'].map(mean_price)
```

```python
df #price got replaced with the mean price of each city. This is target guided ordinal encoding.
```

```python
df[['price', 'city_encoded']]
```
