```python
#Data Encoding - Converting categorical variables into numerical value
#1- Nominal/OHE (One Hot Encoding)- One Hot Encoding is a method to convert categorical variables into a form that could be provided to ML algorithms to do a better job in prediction.
#  It creates new binary columns, each representing a single category from the original categorical variable.
#2-Label and Ordinal Encoding- Label Encoding is a method of converting categorical variables into numerical values by assigning a unique integer to each category.
#  Ordinal Encoding is similar to Label Encoding, but it takes into account the order of the categories, assigning integers based on their rank or order.
#3- Target Guided Ordinal Encoding- Target Guided Ordinal Encoding is a method of encoding categorical variables based on the relationship between the categories and the target variable.
#  It assigns numerical values to categories based on their impact on the target variable, often using statistical measures like mean or median.
```

```python
import pandas as pd
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
#Create a simpp+le dataframe
df = pd.DataFrame({
    'Color': ['Red', 'Blue', 'Green','Green', 'Red', 'Blue'] })
```

```python
df.head()
```

```python
##create an instance of OneHotEncoder
encoder = OneHotEncoder()
#perform fit and transform
encoded=encoder.fit_transform(df[['Color']]).toarray()
#output ids in alphabetical order of the categories- blue first then green and then red
```

```python
import pandas as pd
encoder_df=pd.DataFrame(encoded, columns=encoder.get_feature_names_out(['Color']))
encoder_df
```

```python
#for new data
encoder.transform([['Blue']]).toarray()
```

```python
pd.concat([df, encoder_df], axis=1)
```
