```python
#handling imbalanced datasets
#we use two techniques
#1- Upsampling
#2- Downsampling
```

```python
#creating sample
import numpy as np
import pandas as pd

#set the random seed for reproducibility
np.random.seed(123)

#create a dataframe with 2 classes
n_samples=1000
class_0_ratio=0.9
n_class_0=int(n_samples*class_0_ratio) #900 datapts
n_class_1= n_samples - n_class_0 #1000-900=100 datapts
```

```python
n_class_0,n_class_1
```

```python
#creating dataframe with imbalanced dataset

class_0 = pd.DataFrame({'feature1':np.random.normal(loc=0.0,scale=1.0,size=n_class_0),
                        'feature2':np.random.normal(loc=0.0,scale=1.0,size=n_class_0),
                        'target':[0]*n_class_0})
class_1 = pd.DataFrame({'feature1':np.random.normal(loc=2.0,scale=1.0,size=n_class_1),
                        'feature2':np.random.normal(loc=2.0,scale=1.0,size=n_class_1),
                        'target':[1]*n_class_1})                        
```

```python
df=pd.concat([class_0,class_1]).reset_index(drop=True)
```

```python
df.head()
```

```python
df.tail()
```

```python
df['target'].value_counts()
```

```python
#Upsampling the minority class
df_minority=df[df['target'] == 1]
df_majority=df[df['target'] == 0]
```

```python
from sklearn.utils import resample #it tries to create new points to equalize with respect the majority
df_minority_upsampled = resample(df_minority, replace=True,
n_samples=len(df_majority), random_state=142)
```

```python
df_minority_upsampled.shape
```

```python
df_minority_upsampled.head()  #all samples as 1
```

```python
df_upsampled=pd.concat([df_majority,df_minority_upsampled])
```

```python
df_upsampled  #total 1800 samples, 900 of class 0 and 900 of class 1
```

```python
df_upsampled['target'].value_counts()
```

```python
#downsampling

#creating sample
import numpy as np
import pandas as pd

#set the random seed for reproducibility
np.random.seed(123)

#create a dataframe with 2 classes
n_samples=1000
class_0_ratio=0.9
n_class_0=int(n_samples*class_0_ratio) #900 datapts
n_class_1= n_samples - n_class_0 #1000-900=100 datapts


#creating dataframe with imbalanced dataset

class_0 = pd.DataFrame({'feature1':np.random.normal(loc=0.0,scale=1.0,size=n_class_0),
                        'feature2':np.random.normal(loc=0.0,scale=1.0,size=n_class_0),
                        'target':[0]*n_class_0})
class_1 = pd.DataFrame({'feature1':np.random.normal(loc=2.0,scale=1.0,size=n_class_1),
                        'feature2':np.random.normal(loc=2.0,scale=1.0,size=n_class_1),
                        'target':[1]*n_class_1})    

df=pd.concat([class_0,class_1]).reset_index(drop=True)

#check the distribution of classes
print(df['target'].value_counts())
```

```python
#downsampling the majority class
df_minority=df[df['target'] == 1]
df_majority=df[df['target'] == 0]
```

```python
from sklearn.utils import resample #it tries to create new points to equalize with respect the majority
df_majority_downsampled = resample(df_majority, replace=False,
n_samples=len(df_minority), random_state=142)
```

```python
df_majority_downsampled.shape   #size got reduced to 100 samples
```

```python
df_downsampled=pd.concat([df_minority,df_majority_downsampled])
```

```python
df_downsampled.target.value_counts()  #both classes have equal number of samples
```
