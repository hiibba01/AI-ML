```python
#Create a simpp+le dataframe
import pandas as pd
df = pd.DataFrame({
    'Color': ['Red', 'Blue', 'Green','Green', 'Red', 'Blue'] })
```

```python
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder
lbl_encoder = LabelEncoder()
```

```python
lbl_encoder.fit_transform(df[['Color']])
```

```python
lbl_encoder.transform([['Red']])
```

```python
lbl_encoder.transform([['Blue']])
#here values are assigned in alphabetical order, so Blue is 0, Green is 1 and Red is 2 or in the order of frequency
```

```python
lbl_encoder.transform([['Green']])
```

```python
##ordinal encoding- It is used to convert categorical values into numerical values based on the order of the categories.
##For example, if we have a column with values "Low", "Medium", and "High", we can assign numerical values 0, 1, and 2 respectively to represent the order of the categories.
## This is useful when the categorical values have a natural order or ranking.
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
#create a simple dataframe with ordinal values
df = pd.DataFrame({
    'Size': ['Small', 'Medium', 'Large', 'Medium', 'Small', 'Large'] })
df
```

```python
##create an instance of the OrdinalEncoder class and fit it to the data
encoder=OrdinalEncoder(categories=[['Small', 'Medium', 'Large']])
```

```python
encoder.fit_transform(df[['Size']])
```

```python
encoder.transform([['Small']])
```
